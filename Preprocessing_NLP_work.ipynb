```python

import re, string
import pandas as pd
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
data = pd.read_csv("twitter_training.csv", header=None,
                   names=["id", "topic", "sentiment", "reviews"])
data = data.drop(["id", "topic"], axis=1)
data.drop_duplicates(inplace=True)
data.dropna(inplace=True)
```

```python
stop_set = set(stopwords.words("english"))

def text_preprocessor(text):
    text = text.lower()
    text = re.sub(r"\s+", " ", text)
    text = re.sub(r"\d+", " ", text)
    text = "".join([a for a in text if a not in string.punctuation])
    text = " ".join([a for a in text.split() if a not in stop_set])
    return text

data["reviews"] = data["reviews"].apply(text_preprocessor)
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tok = Tokenizer()
tok.fit_on_texts(data["reviews"])
data["reviews"] = tok.texts_to_sequences(data["reviews"])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
X = pad_sequences(data["reviews"], maxlen=50, padding="post")
```

```python
from sklearn.preprocessing import LabelEncoder
y = LabelEncoder().fit_transform(data["sentiment"])
```
